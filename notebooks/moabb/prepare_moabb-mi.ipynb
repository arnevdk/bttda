```python
from moabb.paradigms import FilterBankMotorImagery
from moabb.datasets import *

paradigm = FilterBankMotorImagery(resample=250)
datasets = paradigm.datasets
exclude = []
datasets = [d for d in datasets if d.code not in exclude and not d.code.startswith('Fake')]
for dataset in datasets:
    print(dataset.code)
```

```python
for dataset in datasets:
    print(dataset.code)
    dataset.download(verbose=True, accept=True)
```

```python
cache_config = dict(
    use=True,
    save_raw=False,
    save_epochs=False,
    save_array=True,
    overwrite_raw=False,
    overwrite_epochs=False,
    overwrite_array=False,
)
for dataset in datasets:
    n_subjects = len(dataset.subject_list)
    for subject in dataset.subject_list:
        print(f'Processing {dataset.code} subject {subject}/{n_subjects} ...')
        paradigm.get_data(dataset, subjects=[subject], cache_config=cache_config)
    print()
```
