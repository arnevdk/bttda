```python
import plotly.io as pio
pio.renderers.default = 'iframe'
```

```python
import os
import pandas as pd

base_path = '../data/results/gridsearch'

results = []
paths = os.listdir(base_path)
for path in paths:
    results.append(pd.read_csv(os.path.join(base_path, path)))
results = pd.concat(results, ignore_index=True)
results
```

```python
import plotly.express as px

df = results.copy()
idx = ['dataset', 'subject', 'fold', 'theta']
df = df.set_index(idx, drop=True)
df = df.sort_index()
first_block = df[df.n_blocks == 1]
metrics = ['train_score', 'test_score', 'train_mse', 'test_mse']
df[[m+'_diff' for m in metrics]] = df[metrics]-first_block[metrics]
df = df.reset_index()
df
```

```python
df= df.query('subject==2')
df = df.groupby(['n_blocks', 'theta'])['test_score'].aggregate('mean').reset_index()
n_colors = df['theta'].nunique()
colors = px.colors.sample_colorscale("viridis", [n/(n_colors -1) for n in range(n_colors)])
fig = px.line(df, x='n_blocks', y='test_score', color='theta', color_discrete_sequence=colors)
fig.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

thetas = [0, 0.1, 1]
df = results[results['theta'].isin(thetas)].copy()
sns.lineplot(df, x='n_blocks', y='test_score', hue='theta', errorbar='ci')
plt.show()
```
