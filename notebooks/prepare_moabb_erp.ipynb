```python
from moabb.paradigms import P300
from moabb.datasets import *

paradigm = P300(resample=48)
datasets = paradigm.datasets

exclude = [
    # Did not download
    'ErpCore2021-N2pc',
    'Sosulski2019',
    # Not included
    'DemonsP300',
    'FakeVirtualRealityDataset-p300-21-1--60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60-60--120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120-120--target-nontarget--c3-cz-c4',
]
datasets = [d for d in datasets if d.code not in exclude and not d.code.startswith('Fake')]
datasets = [Lee2019_ERP()]
for dataset in datasets:
    print(dataset.code)
```

```python
from tqdm import tqdm

cache_config = dict(
    use=True,
    save_raw=False,
    save_epochs=False,
    save_array=True,
    overwrite_raw=False,
    overwrite_epochs=False,
    overwrite_array=False,
)
for dataset in datasets:
    n_subjects = len(dataset.subject_list)
    for si, subject in tqdm(list(enumerate(dataset.subject_list)), desc=f'Processing {dataset.code}'):
        try:
            paradigm.get_data(dataset, subjects=[subject], cache_config=cache_config)
        except Exception as e:
            print(e)
    print()
```
