```python
from moabb.paradigms import MotorImagery
from moabb.datasets import *

sfreq=250
n_classes=3
paradigm = MotorImagery(resample=sfreq, n_classes=n_classes)

datasets = paradigm.datasets
"""
datasets = [
    AlexMI(),
    BNCI2014_001(),
    Zhou2016(),
    Weibo2014(),
    Schirrmeister2017()
]"""
datasets = [AlexMI()]
for dataset in datasets:
    print(dataset.code)
    
```

```python
from sklearn.preprocessing import FunctionTransformer
from classification_mi import stf_transform

cache_config = dict(
    use=True,
    save_raw=False,
    save_epochs=False,
    save_array=True,
    overwrite_raw=False,
    overwrite_epochs=False,
    overwrite_array=True,
)
for dataset in datasets:
    print(f'dataset: {dataset.code}, classes: {dataset.event_id}')
    n_subjects = len(dataset.subject_list)
    events = list(dataset.event_id.keys())[:n_classes]
    paradigm = MotorImagery(events=events, n_classes=n_classes, resample=sfreq)
    for subject in dataset.subject_list:
        print(f'Processing {dataset.code} subject {subject}/{n_subjects} ...')
        X,y,meta = paradigm.get_data(
            dataset,
            subjects=[subject],
            cache_config=cache_config,
            postprocess_pipeline = FunctionTransformer(stf_transform)
 
        )
```
